# GitHub

## 1.1 What is GitHub?

GitHub is a web-based hosting platform for Git repositories. It stores repositories in the cloud and adds a collaboration layer on top of Git: issues, pull requests, code review, discussions, wikis, project boards, CI/CD (GitHub Actions) and access control. It is where most open source projects live and where teams work on code together.

## 1.2 How is GitHub related to Git?

Git is the version control system itself: a command line tool that runs locally and tracks changes to files. GitHub is a service that hosts Git repositories remotely.

- Git works without GitHub.
- GitHub cannot exist without Git.
- GitHub is one of many hosts (others: GitLab, Bitbucket).

Git is the engine, GitHub is the garage and the social network around it.

## 1.3 How does forking a repository differ from cloning it?

**Clone** copies a repository from a remote server to MY **local machine**:

```bash
git clone <url>
```

I get a local copy. If I don not have write access to the original repository, I cannot push to it.

**Fork** creates a copy of someone else's repository **under MY own GitHub account** (server side, done through the GitHub UI). I have full write access to your fork.

Standard open source flow:

1. Fork the upstream repository.
2. Clone **MY fork** locally.
3. Create a branch and commit MY changes.
4. Push the branch to MY fork.
5. Open a pull request from MY fork to the upstream repository.

## 1.4 Why are pull requests needed and how do they work?

A pull request (PR) proposes that your changes be merged into another branch or repository. It exists so that changes can be **reviewed, discussed and tested** before entering the main codebase, instead of anyone pushing directly to `main`.

How it works:

1. Create a branch and commit your changes.
2. Push the branch to GitHub.
3. Open a PR: choose the base branch (for example `upstream/main`) and the compare branch (my branch).
4. GitHub shows a diff, and automated checks (CI, linters) run.
5. Reviewers comment, request changes or approve.
6. I push more commits to the same branch and the PR updates automatically.
7. A maintainer merges it.

If the PR description contains `Closes <issue-url>`, the linked issue is closed automatically on merge.

## 1.5 Does GitHub use your email address to link your Git commits to your account?

Yes. Every Git commit records an author email taken from `git config user.email`. GitHub matches that email against the verified emails on your account.

- If it matches, the commit is attributed to your profile and counts toward my contribution graph.
- If it does not match, the commit still appears in the history but is not linked to my account: no avatar, no contribution credit.

```bash
git config user.email
```

The value must be one of the emails added to your GitHub account settings.

## 1.6 Which command generates an SSH key for SSH access to a repository?

```bash
ssh-keygen -t ed25519 -C "your_email@example.com"
```

Legacy alternative:

```bash
ssh-keygen -t rsa -b 4096 -C "your_email@example.com"
```

This creates a private key (`~/.ssh/id_ed25519`) and a public key (`~/.ssh/id_ed25519.pub`). The **public** key is then added to GitHub under Settings -> SSH and GPG keys.
